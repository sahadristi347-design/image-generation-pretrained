```python
# Import PyTorch for deep learning and StableDiffusionPipeline to load the model
import torch
from diffusers import StableDiffusionPipeline
```

```python
# Define the pre-trained model ID from Hugging Face
model_id = "runwayml/stable-diffusion-v1-5"

# Automatically check if a GPU is available
device = "cuda" if torch.cuda.is_available() else "cpu"

# Load the pre-trained model into the pipeline with half-precision for faster generation
pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16 if device == "cuda" else torch.float32
)

# Move the model to the GPU hardware
pipe = pipe.to(device)
```

```python
# Write your text prompt describing the image you want to create (Change this text anytime!)
prompt = "A cute futuristic cyberpunk cat sitting on a neon-lit rooftop, highly detailed, digital art"

# Generate the image from the text prompt
image = pipe(prompt).images[0]

# Display the generated image directly inside Google Colab
image
```

```python
# Save the generated image locally in your Colab session
image.save("generated_image.png")

print("Image saved successfully as 'generated_image.png'!")
```
